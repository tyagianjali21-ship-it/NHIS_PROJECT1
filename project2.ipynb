#Outlier Detection:
There are 3 main tpyes of outlier detection methods:
1. Using Boxplot (Visual Detection)
2. Using IQR method
3. Using Z-score method

```python
#Boxplot Method
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=['annual_revenue_million'])
plt.show
```

```python
#IQR Method:
Q1=df['annual_revenue_million'].quantile(0.25)
Q3=df['annual_revenue_million'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
  (df['annual_revenue_million']<
 lower_bound) |
  (df['annual_revenue_million']>
 upper_bound)]
print(outliers)
```

```python
#Using Zscore:
from scipy.stats import zscore

z_scores = zscore(df['annual_revenue_million'])
outliers = df[abs(z_scores) > 3]
print(outliers)
```

#Insights:
There are no outliers even after checking it with all 3 methods.

```python
1. pair plot method:

import seaborn as sns 

sns.pairplot(df[['employee_count','annual_revenue_million', 'profit_margin_percent']])
```

```python
#CorrelationHeatmap:

import seaborn as sns
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=True,
cmap='coolwarm')
plt.show()
        
```

```python
#Insigts:
Most correlation values are very close to 0.
This means very weak relationsip between variables.
No pair shows strong positive correlation or strong negative correlation.
The diagonal values are 1 because each variable is perfectly correlated with itself.
```
